```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the Titanic dataset
df = sns.load_dataset("titanic")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Save the Titanic dataset
df.to_csv("titanic.csv", index=False)

print("Titanic dataset saved successfully!")
```

```python
# Basic dataset inspection

print("DATASET SHAPE")
print(df.shape)

print("\nDATASET INFO")
df.info()

print("\nDESCRIPTIVE STATISTICS")
display(df.describe(include="all"))
```

```python
# Calculate missing values and missing percentages

missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

# Show only columns that contain missing values
missing_summary = missing_summary[missing_summary["Missing Count"] > 0]

display(missing_summary)
```

```python
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
]

display(missing_summary)
```

```python
# Handle missing values according to the assignment thresholds

# 1. Age: 5–30% missing → median imputation
age_median = df["age"].median()
df["age"] = df["age"].fillna(age_median)

# 2. Embarked: <5% missing → drop affected rows
df = df.dropna(subset=["embarked"])

# 3. Embark_town: <5% missing → drop affected rows
df = df.dropna(subset=["embark_town"])

# 4. Deck: >30% missing → drop the column
df = df.drop(columns=["deck"])

print("Missing-value treatment completed.")
print("New shape:", df.shape)
```

```python
print("Remaining missing values:")
display(df.isnull().sum())
```

```python
print(df["age"].describe())
print("\nAge histogram data:")
print(pd.cut(df["age"], bins=10).value_counts().sort_index())
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0]

print("Missing values:")
print(missing)

print("\nMissing percentage:")
print((missing / len(df) * 100).round(2))
```

```python
print("Fare Mean:", df["fare"].mean())
print("Fare Median:", df["fare"].median())
print("Fare Mode:", df["fare"].mode()[0])

if df["fare"].mean() > df["fare"].median() > df["fare"].mode()[0]:
    print("Fare distribution: Right-skewed")
elif df["fare"].mean() < df["fare"].median() < df["fare"].mode()[0]:
    print("Fare distribution: Left-skewed")
else:
    print("Fare distribution: Approximately symmetric")
```

```python
print("Survival rate by SEX:")
print(df.groupby("sex")["survived"].mean())

print("\nSurvival rate by PCLASS:")
print(df.groupby("pclass")["survived"].mean())

print("\nSurvival rate by SEX and PCLASS:")
print(df.groupby(["sex", "pclass"])["survived"].mean())
```
